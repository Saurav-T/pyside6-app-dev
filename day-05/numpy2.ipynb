### Core Operations

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print(a + b) # Element Wise Operations
```

```python
print("Standard Deviation:", a.std())
print("Variance:", a.var())
```

2. Slicing Mastery<br>
arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])<br>
Extract:
- First Row
- Last Row
- Second Column
- Last Column
- Submatrix: [[20 30]
 [50 60]]

```python
arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

print("First Row:", arr[0])
print("Last Row:", arr[-1])
print("Second Column:", arr[[0, 1, 2],[1, 1, 1]])
print("Last Column:", arr[[0, 1, 2], [-1, -1, -1]])
print("Sub Matrix:", arr[0:2, 1:3])
```

3. Boolean Filtering<br>
scores = np.array([45, 67, 82, 31, 95, 58, 77])<br>
Tasks:
- Extract scores above 60
- Extract scores between 50 and 80
- Count scores below 50
- Replace all failing marks (<40) with 0

```python
scores = np.array([45, 67, 82, 31, 95, 58, 77])

print("Scores above 60:", scores[scores > 60] )
print("Scores between 50 and 80:", scores[(scores > 50) & (scores < 80)])
print("Count scores below 50:", scores[scores < 60].size)

scores[scores < 40] = 0
print("Replace failing marks:", scores)
```

4. Broadcasting and Arithmetic<br>
prices = np.array([100, 250, 80, 300])<br>
Compute:
- Price after 13% tax
- Price after Rs. 20 discount
- Double every price
- Find items costing more than Rs. 200


```python
prices = np.array([100, 250, 80, 300])

print("Price after 13% tax:", prices + (prices * 0.13))
print("Price after Rs. 20 discount:", prices - 20)
print("Double prices:", prices * 2)
print("Items costing more than 200:", prices[prices > 200])

```

5. Aggregations<br>
sales = np.array([
    [100, 120, 140],
    [90, 110, 150],
    [80, 130, 170]
])<br>
Find:
- Total sales
- Average sale
- Maximum sale
- Sum of each row
- Sum of each column
- Average of each column

```python
sales = np.array([
    [100, 120, 140],
    [90, 110, 150],
    [80, 130, 170]
])

print("Total sales:", sales.sum())
print("Average sale:", sales.sum() / sales.flatten().size)
print("Maximum sale:", sales.max())
print("Sum of each row:", sales.sum(axis=0))
print("Sum of each column:", sales.sum(axis=1))
print("Average of each column:", sales.sum(axis=1)/sales.shape[1])
```
