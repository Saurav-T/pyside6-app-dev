# Numpy Tutorials

### Core Operations

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print(a + b) # Element Wise Operations
```

```python
a = np.array([1, 2, 3])
print(a + 5) # Broadcasting
```

```python
a = np.array([
    [1,2,3],
    [4,5,6]
])

b = np.array([10,20,30]) # Row Broadcasting

print(a + b)
```

```python
a = np.array([
    [1,2,3],
    [4,5,6]
])

b = np.array([
    [10],
    [20]
])

print(a + b) # Column Broadcasting
```

```python
a = np.array([10, 20, 30, 40])
print("Sum:", a.sum())
print("Mean:", a.mean())
print("Max:", a.max())
print("Min:", a.min())
print("Size:", a.size)
```

```python
print("Standard Deviation:", a.std())
print("Variance:", a.var())
```

```python
a = np.array([5, 12, 18, 3, 25])

(a > 10).sum()
```

```python
a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

a.sum(axis=0)
```

```python
a = np.array([10, 50, 30, 20])

print(a.argmax()) # Index
print(a.argmin())
```

```python
a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)

print(b)
```

```python
print(b.flatten())
```

```python
print(b.T) # Transpose
```

```python
a = np.array([[[1, 2, 3]]])

print(a.shape)
print(a.squeeze().shape) # Remove Single Dimensions
```
