# Numpy Tutorials

### Intermediate Power Tools

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.vstack((a, b)))
```

```python
print(np.hstack((a, b)))
```

```python
print(np.concatenate((a, b), axis=0))
```

```python
a = np.array([1, 2, 3, 4, 5, 6])

print(np.split(a, 3))
```

```python
a = np.array([1, 2, 3, 4, 5])

print(np.array_split(a, 2))
```

```python
a = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])

np.split(a, 2, axis=0)
```

```python
np.split(a, 2, axis=1)
```

### Excercises

1. Tasks:
- Stack vertically using vstack
- Stack horizontally using hstack
- Try the same using concatenate

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("V Stack:\n", np.vstack((a, b)))
print("H Stack:\n", np.hstack((a, b)))
print("Concatenate(Horizontal):\n", np.concatenate((a, b), axis=0))
```

2. Tasks:
- Concatenate with axis=0
- Concatenate with axis=1

```python
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

print(np.concatenate((a, b), axis=0))
print(":::::::::::::::::::::")
print(np.concatenate((a, b), axis=1))
```

3. Tasks:
- Split into 2 equal parts (rows)
- Split into 3 column groups using array_split

```python
data = np.array([
    [1, 2, 9],
    [3, 4, 10],
    [5, 6, 11],
    [7, 8, 12]
])

print(np.split(data, 2))
print(np.array_split(data, 3))
```
