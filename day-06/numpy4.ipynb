# Numpy Tutorials

### Linear Algebra

```python
import numpy as np

a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])
```

```python
a @ b # Matrix multiplication OR
np.dot(a, b)
```

```python
a.T # Transpose
```

```python
np.eye(3) # Identity Matrix
```

```python
np.linalg.det(a) # Determinant
```

```python
np.linalg.inv(a) # Inverse
```

```python
A = np.array([[2, 1],
              [1, 3]])

B = np.array([8, 13])
```

```python
x = np.linalg.solve(A, B) # Ax = B, We're finding x here
print(x)
```

```python
Q, R = np.linalg.qr(a)
print("Orthogonal matrix Q:")
print(Q)
print("Upper triangular matrix R:")
print(R)
```

```python
U, S, Vt = np.linalg.svd(a)
print("Left singular vectors (U):")
print(U)
print("Singular values (S):")
print(S)
print("Right singular vectors (V^T):")
print(Vt) 
```

```python
values, vectors = np.linalg.eig(a)
print("Eigenvalues:")
print(values)
print("Eigenvectors:")
print(vectors)
```

3. Solving system<br>
2x + y = 7<br>
x + 3y = 11<br>

```python
A = [[2, 1],
     [1, 3]]
b = [7, 11]

solution = np.linalg.solve(A, b)
solution
```

4. Eigenvalues, Eigenvectors and Norm

```python
A = np.array([
    [4, 2],
    [1, 3]
])
```

```python
values, vectors = np.linalg.eig(A)
values
```

```python
vectors
```

```python
v = vectors[:, 0]
v
```

```python
A @ v
```

```python
values[0] * v
```

```python
np.linalg.norm(v)
```
