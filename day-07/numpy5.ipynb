### Structured Arrays

```python
students = np.array([
    ("Alice", 20, 3.8),
    ("Bob", 22, 3.5),
    ("Charlie", 21, 3.9)
], dtype=[
    ("name", "U10"),
    ("age", "i4"),
    ("gpa", "f4")
])
```

```python
students.dtype
```

```python
students["age"]
```

```python
students[0]["name"]
```

```python
students[students["age"] > 20]
```

### Excercises

1. Tasks:
- Compute square of all elements using:
    - Python loop
    - NumPy vectorization
- Compare execution time (use %timeit in Jupyter or manual timing)
- Compute:
    - sum of elements > 50
    - without loops

```python
a = np.random.randint(1, 100, size=10_000)
a
```

```python
%%timeit
a = np.random.randint(1, 100, size=10_000)
for i, val in enumerate(a):
    a[i] = val * val
print(a)
```

```python
%%timeit
a * a
```

```python
%%timeit
sum = 0
for val in a[a > 50]:
    sum += val

print(sum)
```

```python
%%timeit
np.sum(a[a > 50])
```
