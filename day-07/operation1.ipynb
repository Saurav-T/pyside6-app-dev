# Advance operation with business examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure: [restaurant_id, 2021, 2022, 2023, 2024]

data = np.array([
    [1, 12000, 13500, 15000, 16000],
    [2,  8000,  9500, 11000, 12500],
    [3, 20000, 22000, 24000, 26000],
    [4,  5000,  7000,  6500,  9000],
    [5, 15000, 15500, 17000, 18000]
])
```

```python
print(":::::: Sales Analysis ::::::")
print("\n Sales Data Shape:", data.shape)
print("\n Sample Data for 1st 3 Restaurants:\n", data[:, 1:])
```

```python
print("Total Sales per year:", np.sum(data[:,1:], axis=0))
print("Total Sales of a Restaurant:", np.sum(data[:,1:], axis=1))
```

```python
print("Minimum Sales of each restaurant:", np.min(data[:, 1:],  axis=1))
print("Minimum Sales each year:", np.min(data[:, 1:],  axis=0))
```

```python
print("Maximum Sales of each restaurant:", np.max(data[:, 1:],  axis=1))
print("Maximum Sales each year:", np.max(data[:, 1:],  axis=0))
```

```python
print("Average Sales of each restaurant:", np.mean(data[:, 1:],  axis=1))
```

```python
print("Cumulative Sales of each restaurant:\n", np.cumsum(data[:, 1:],  axis=1))
```

```python
plt.figure(figsize=(10, 6))
plt.plot(np.mean(np.cumsum(data[:, 1:],  axis=0), axis=0))
plt.title("Average cumulative sales across all restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
