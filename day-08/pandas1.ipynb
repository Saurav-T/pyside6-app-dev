### Pandas Basics

#### Series

```python
import pandas as pd

x = pd.Series([10, 20, 30, 40, 50]) # List
```

```python
y = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']) # Custom Index
```

```python
z = pd.Series({'a': 1, 'b': 2, 'c': 3}) # Dictionary
```

```python
x[:2] # Slicing
```

```python
y + 5 # Operations
```

```python
print("Index of x:", x.index)
print("Index of y:", y.index)
```

```python
x.values
```

```python
y.dtype
```

#### Dataframe

```python
data = {
    "Name": ["Aarav", "Sita", "Rohan"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)
print(df)
```

```python
df["Name"]
df[["Name", "Marks"]]
```

```python
df.loc[0] # Label Based
df.iloc[0] # Position Based
```

```python
df["Marks"].mean()
```

```python
print(df.shape)
print(df.columns)
print(df.index)
print(df.dtypes)
```
