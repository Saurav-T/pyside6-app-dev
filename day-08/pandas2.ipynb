```python
data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Aarav", "Sita", "Rohan", "Meera", "Anil"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 92, 78, 88, 95],
    "City": ["Kathmandu", "Lalitpur", "Bhaktapur", "Kathmandu", "Pokhara"],
    "Gender": ["M", "F", "M", "F", "M"]
}

df = pd.DataFrame(data)
```

```python
df["Marks"].apply(lambda x: x + 5)
```

```python
df.apply(lambda row: row["Marks"] + row["Age"], axis=1)
```

```python
df["Gender"].map({"M": "Male", "F": "Female"})
```

```python
df["Percentage"] = (df["Marks"] / 100) * 100
```

```python
df
```

```python
df["Result"] = np.where(df["Marks"] >= 40, "Pass", "Fail")
```

```python
df
```

```python
dates = pd.DataFrame({
    "Date": ["2024-01-01", "2024-02-15", "2024-03-10"]
})

dates["Date"] = pd.to_datetime(dates["Date"])
```

```python
dates["Year"] = dates["Date"].dt.year
dates["Weekday"] = dates["Date"].dt.day_name()
dates
```
