# Pandas Tutorial

### Grouping and Aggregation

```python
import pandas as pd

data = {
    "Region": ["East", "East", "West", "West", "North"],
    "Category": ["Tech", "Furniture", "Tech", "Furniture", "Tech"],
    "Month": ["Jan", "Jan", "Feb", "Feb", "Jan"],
    "Sales": [500, 300, 700, 200, 400]
}

df = pd.DataFrame(data)
```

```python
df.groupby("Region")["Sales"].sum()
```

```python
df.groupby(["Region", "Category"])["Sales"].sum()
```

```python
df.groupby("Category")["Sales"].mean()
```

```python
df.groupby("Region").agg({
    "Sales": ["sum", "mean"],
})
```

```python
df.pivot_table(
    values="Sales",
    index="Region",
    columns="Category",
    aggfunc="sum"
)
```
