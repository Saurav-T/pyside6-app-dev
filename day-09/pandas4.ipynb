### Data Merging

#### Joins

```python
import pandas as pd

df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Aarav", "Sita", "Rohan"]
})

df2 = pd.DataFrame({
    "ID": [1, 2, 4],
    "Marks": [85, 90, 78]
})
```

```python
pd.merge(df1, df2, on="ID", how="inner")
```

```python
pd.concat([df1, df2], axis=1)
```

```python
df1["ID"].is_unique
```

```python
df1 = df1.drop_duplicates(subset="ID")
df1
```
