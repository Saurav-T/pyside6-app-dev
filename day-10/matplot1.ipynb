# Matplotlib Tutorial

```python
import matplotlib.pyplot as plt

fig = plt.figure()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.plot([1, 2, 3], [4, 5, 6])

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.plot([1, 2, 3, 4], [1, 4, 9, 16])

ax.set_title("Square Numbers")
ax.set_xlabel("X")
ax.set_ylabel("Y")

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2, 2)

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2, 2)

axs[0,0].plot([1,2,3], [1,4,9])
axs[0,1].bar(["A","B"], [5,3])
axs[1,0].scatter([1,2,3], [3,2,5])
axs[1,1].hist([1,2,2,3,3,3])

plt.show()
```

```python
import matplotlib.pyplot as plt

# Line Plot
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 20, 25]

plt.plot(x, y)

plt.title("Sales Growth")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 120, 150, 180]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```
