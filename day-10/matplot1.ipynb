# Matplotlib Tutorial

### Figures and Axes

```python
import matplotlib.pyplot as plt

fig = plt.figure()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.plot([1, 2, 3], [4, 5, 6])

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.plot([1, 2, 3, 4], [1, 4, 9, 16])

ax.set_title("Square Numbers")
ax.set_xlabel("X")
ax.set_ylabel("Y")

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2, 2)

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2, 2)

axs[0,0].plot([1,2,3], [1,4,9])
axs[0,1].bar(["A","B"], [5,3])
axs[1,0].scatter([1,2,3], [3,2,5])
axs[1,1].hist([1,2,2,3,3,3])

plt.show()
```

### Line Plot

```python
import matplotlib.pyplot as plt

# Line Plot
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 20, 25]

plt.plot(x, y)

plt.title("Sales Growth")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 120, 150, 180]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

### Scatter Plot

```python
import matplotlib.pyplot as plt

# Scatter Plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]

plt.scatter(x, y)

plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
scores = [45, 50, 60, 65, 80, 90]

plt.scatter(study_hours, scores)

plt.title("Study Hours vs Score")
plt.xlabel("Study Hours")
plt.ylabel("Score")

plt.show()
```

### Bar Chart

```python
import matplotlib.pyplot as plt

students = ["Alice", "Bob", "Charlie", "David", "Eva"]
scores = [78, 85, 62, 90, 74]

plt.bar(students, scores)

plt.title("Student Scores")
plt.xlabel("Students")
plt.ylabel("Scores")

plt.show()
```

```python
plt.barh(students, scores)
```

### Histogram

```python
import matplotlib.pyplot as plt

data = [45, 50, 52, 55, 60, 61, 62, 65, 70, 72, 75, 80]

plt.hist(data, bins=5)

plt.title("Score Distribution")
plt.xlabel("Scores")
plt.ylabel("Frequency")

plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ["Python", "Java", "C++", "JavaScript"]
sizes = [40, 25, 15, 20]

plt.pie(sizes, labels=labels, autopct="%1.1f%%")

plt.title("Programming Language Preference")

plt.show()
```

### Box Plot

```python
import matplotlib.pyplot as plt

data = [12, 15, 14, 10, 18, 20, 22, 25, 30, 35, 40]

plt.boxplot(data)

plt.title("Data Distribution")
plt.ylabel("Values")

plt.show()
```

```python
import matplotlib.pyplot as plt

class_A = [10, 12, 14, 15, 18]
class_B = [20, 22, 25, 28, 30]

plt.boxplot([class_A, class_B], labels=["Class A", "Class B"])

plt.title("Score Comparison")
plt.ylabel("Scores")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5, 6, 7])
y1 = np.array([10, 50, 100, 500, 1000, 5000, 10000])
y2 = np.array([5, 20, 60, 200, 800, 3000, 9000])

plt.figure(figsize=(8, 5))

plt.plot(x, y1, "o--", label="Dataset 1")
plt.plot(x, y2, "s-", label="Dataset 2")

plt.title("Log Scale Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.legend()
plt.grid(True, which="both", linestyle="--", alpha=0.5)

plt.yscale("log")

plt.show()
```
