```python
import pandas as pd
```

```python
sales = pd.Series([10, 20, 30, 40])
sales.rolling(window=3).mean()
```

```python
print(sales.rolling(window=3).sum())
print(sales.rolling(window=3).max())
print(sales.rolling(window=3).min())
```

```python
sales.expanding().mean()
```

```python
import numpy as np

data = {
    "ID": [1, 2, 3, 4, 5, 6],
    "Name": ["Aarav", "Sita", "Rohan", "Meera", "Anil", "Kiran"],
    "Age": [20, 21, 19, 22, 20, 23],
    "Marks": [85, 92, 78, 88, 95, 60],
    "City": ["Kathmandu", "Lalitpur", "Pokhara", "Kathmandu", "Butwal", "Pokhara"],
    "Gender": ["M", "F", "M", "F", "M", "M"],
    "Join_Date": ["2024-01-01", "2024-02-15", "2024-03-10", "2024-01-20", "2024-05-01", "2024-06-01"]
}

df = pd.DataFrame(data)
```

```python
df.query("Marks > 80")
```

```python
df.query("Marks > 80 and Age < 22")
```

```python
df.query("Marks > 90 or Age == 19")
```

```python
min_marks = 80

df.query("Marks > @min_marks")
```

```python
df.query("Name == 'Aarav'")
```
