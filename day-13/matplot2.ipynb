# Matplotlib Tutorial

### Anotations and Twin Axes

```python
# Annotation
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 25, 15, 30, 20])

fig, ax = plt.subplots()

ax.plot(x, y, marker="o")

# Annotate a point (highest value)
ax.annotate(
    "Peak Value",
    xy=(4, 30),
    xytext=(4.2, 32),
    arrowprops=dict(facecolor="black", arrowstyle="->"),
    fontsize=10
)

ax.set_title("Annotation Example")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

days = np.array([1, 2, 3, 4, 5, 6, 7])
sales = np.array([100, 150, 130, 200, 250, 220, 300])

fig, ax = plt.subplots()

ax.plot(days, sales, marker="o")

# Find max point
max_index = np.argmax(sales)

ax.annotate(
    "Highest Sales",
    xy=(days[max_index], sales[max_index]),
    xytext=(days[max_index]-1, sales[max_index]+20),
    arrowprops=dict(facecolor="red", arrowstyle="->"),
    fontsize=10,
    color="red"
)

ax.set_title("Daily Sales with Annotation")
ax.set_xlabel("Day")
ax.set_ylabel("Sales")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y1 = np.array([10, 20, 30, 40, 50])   # Dataset 1
y2 = np.array([100, 80, 60, 40, 20])   # Dataset 2

fig, ax1 = plt.subplots()

# First plot (left axis)
ax1.plot(x, y1, "r-o", label="Line 1")
ax1.set_xlabel("X Axis")
ax1.set_ylabel("Y1 (Left Axis)", color="red")
ax1.tick_params(axis="y", labelcolor="red")

# Second axis (right axis)
ax2 = ax1.twinx()
ax2.plot(x, y2, "b-s", label="Line 2")
ax2.set_ylabel("Y2 (Right Axis)", color="blue")
ax2.tick_params(axis="y", labelcolor="blue")

plt.title("Twin Y-Axis Example")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

months = np.array([1, 2, 3, 4, 5])
revenue = np.array([200, 300, 400, 500, 600])
profit_margin = np.array([20, 25, 30, 28, 35])

fig, ax1 = plt.subplots()

# Revenue (left axis)
ax1.plot(months, revenue, "g-o", label="Revenue")
ax1.set_ylabel("Revenue ($)", color="green")
ax1.tick_params(axis="y", labelcolor="green")

# Profit margin (right axis)
ax2 = ax1.twinx()
ax2.plot(months, profit_margin, "b-s", label="Profit Margin")
ax2.set_ylabel("Profit Margin (%)", color="blue")
ax2.tick_params(axis="y", labelcolor="blue")

plt.title("Revenue vs Profit Margin")

plt.show()
```

```python
# Twinx
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.array([1, 2, 3, 4, 5, 6, 7])
temp = np.array([30, 32, 35, 33, 31, 29, 28])
humidity = np.array([60, 65, 70, 75, 80, 78, 72])

# Figure and axis
fig, ax1 = plt.subplots()

# First plot (Temperature)
ax1.plot(x, temp, "r-o", label="Temperature (°C)")
ax1.set_xlabel("Day")
ax1.set_ylabel("Temperature (°C)", color="red")
ax1.tick_params(axis="y", labelcolor="red")

# Annotation (mark highest temperature)
max_index = np.argmax(temp)
ax1.annotate(
    "Peak Temp",
    xy=(x[max_index], temp[max_index]),
    xytext=(x[max_index] + 0.5, temp[max_index] + 1),
    arrowprops=dict(facecolor="red", shrink=0.05),
    fontsize=10,
    color="red"
)

# Second axis (Humidity)
ax2 = ax1.twinx()
ax2.plot(x, humidity, "b--s", label="Humidity (%)")
ax2.set_ylabel("Humidity (%)", color="blue")
ax2.tick_params(axis="y", labelcolor="blue")

# Title
plt.title("Temperature vs Humidity (Twin Axes + Annotation)")

# Grid
ax1.grid(True, linestyle="--", alpha=0.5)

plt.show()
```
