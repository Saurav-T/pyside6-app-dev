# Matplotlib Tutorial

### Themes

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use("ggplot")

x = np.arange(1, 6)
y = x ** 2

plt.plot(x, y, marker="o")

plt.title("Styled Plot (ggplot)")
plt.show()
```

### rcParams

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.linestyle"] = "--"
plt.rcParams["axes.titlesize"] = 16
plt.rcParams["axes.labelsize"] = 12
plt.rcParams["lines.linewidth"] = 2
plt.rcParams["lines.markersize"] = 8

x = np.arange(1, 6)
y = x ** 2

plt.plot(x, y, marker="o")

plt.title("Plot using rcParams")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

### Custom Styles

```python
import matplotlib.pyplot as plt
import numpy as np

# Custom style settings
plt.rcParams.update({
    "figure.figsize": (8, 5),
    "axes.facecolor": "#f5f5f5",
    "axes.edgecolor": "black",
    "axes.grid": True,
    "grid.linestyle": "--",
    "grid.alpha": 0.6,
    "lines.linewidth": 2,
    "lines.markersize": 6,
    "font.size": 12
})

# Data
x = np.arange(1, 6)
y = x ** 2

# Plot
plt.plot(x, y, marker="o", color="purple")

plt.title("Custom Styled Plot")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 6)
y = x ** 2

with plt.style.context({
    "axes.facecolor": "#1e1e1e",
    "figure.facecolor": "#1e1e1e",
    "axes.labelcolor": "white",
    "xtick.color": "white",
    "ytick.color": "white",
    "text.color": "white"
}):
    plt.plot(x, y, marker="o", color="cyan")
    plt.title("Temporary Dark Style")
    plt.show()
```
