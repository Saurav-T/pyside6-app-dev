# Seaborn Tutorial

### Themes

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

sns.set_theme()

x = np.arange(1, 6)
y = x ** 2

plt.plot(x, y)
plt.title("Seaborn Theme Example")
plt.show()
```

### Scatter Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 40, 60]

sns.scatterplot(x=x, y=y)

plt.title("Scatter Plot Example")
plt.show()
```

### Line Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 40, 60]

sns.lineplot(x=x, y=y)

plt.title("Line Plot Example")
plt.show()
```

### Distribution Plots

```python
# Hist Plot
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

sns.set_theme()

data = np.random.randn(1000)

sns.histplot(data, bins=30, kde=False)

plt.title("Histogram using histplot")
plt.show()
```

```python
# KDE Plot
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

sns.set_theme()

data = np.random.randn(1000)

sns.kdeplot(data)

plt.title("KDE Plot")
plt.show()
```

```python
# Dis Plot (Hist + KDE)
import seaborn as sns
import numpy as np

sns.set_theme()

data = np.random.randn(1000)

sns.displot(data, bins=30, kde=True)

plt.title("Displot Example")
```

### Categorical Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()

categories = ["A", "B", "C", "D"]
values = [10, 20, 15, 25]

sns.barplot(x=categories, y=values)

plt.title("Basic Barplot")
plt.show()
```

```python
import seaborn as sns
import pandas as pd

sns.set_theme()

data = pd.DataFrame({
    "group": ["A", "A", "B", "B", "C", "C"],
    "score": [10, 20, 30, 40, 50, 60]
})

sns.barplot(x="group", y="score", data=data)

plt.title("Barplot (Mean per Group)")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()

data = ["A", "A", "B", "B", "B", "C"]

sns.countplot(x=data)

plt.title("Countplot Example")
plt.show()
```

```python
import seaborn as sns
import pandas as pd

sns.set_theme()

df = pd.DataFrame({
    "category": ["A", "A", "B", "B", "B", "C"]
})

sns.countplot(x="category", data=df)

plt.title("Countplot with DataFrame")
plt.show()
```
