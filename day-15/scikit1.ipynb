### Linear Regression

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
# X = input (features)
X = np.array([[1], [2], [3], [4], [5]])

# y = output (target)
y = np.array([2, 4, 6, 8, 10])

# Model
model = LinearRegression()

model.fit(X, y)

y_pred = model.predict([[6], [7]])

print("Predictions:", y_pred)

# Parameters learned
print("Slope (w):", model.coef_)
print("Intercept (b):", model.intercept_)
```

```python
# X = [size, rooms]
X = np.array([
    [1000, 2],
    [1500, 3],
    [2000, 3],
    [2500, 4],
    [3000, 4]
])

# y = price
y = np.array([200, 300, 400, 500, 600])

model = LinearRegression()
model.fit(X, y)

new_house = np.array([[1800, 3]])
prediction = model.predict(new_house)

print("Prediction:", prediction)
print("Coeficient:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

plt.scatter(X, y)          # real points
plt.plot(X, y_pred)        # predicted line

plt.show()
```

```python
X = np.array([
    [1000, 2],
    [1500, 3],
    [2000, 3],
    [2500, 4],
    [3000, 4]
])

y = np.array([200, 300, 400, 500, 600])

model = LinearRegression()
model.fit(X, y)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(X[:,0], X[:,1], y)

# plane prediction points
x1 = X[:,0]
x2 = X[:,1]
y_pred = model.predict(X)

ax.plot_trisurf(x1, x2, y_pred)

plt.show()
```

### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
# Input feature
X = np.array([[1], [2], [3], [4], [5], [6]])

# Labels
y = np.array([0, 0, 0, 1, 1, 1])
```

```python
model = LogisticRegression()

model.fit(X, y)

prediction = model.predict([[3.5]])

print("Prediction:", prediction)
prob = model.predict_proba([[3.5]])

print("Probability:", prob)

print("Weight(W):", model.coef_)
print("Bias(b):", model.intercept_)
```

```python
# Tumor Classification

# Feature: size
X = np.array([[1], [2], [3], [4], [5], [6]])

# Target: benign(0), malignant(1)
y = np.array([0, 0, 0, 1, 1, 1])
```

```python
model = LogisticRegression()

model.fit(X, y)

prediction = model.predict([[3.5]])

print("Prediction:", prediction)

prob = model.predict_proba([[3.5]])

print("Probability:", prob)

sizes = [[2], [3.5], [5]]

print("Multiple Predictions:", model.predict(sizes))

print("Weight(W):", model.coef_)
print("Bias(b):", model.intercept_)
```

### Decision Tree Classification

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
```

```python
X = np.array([[1],
              [2],
              [3],
              [4],
              [5],
              [6]])

y = np.array([0, 0, 0, 1, 1, 1])

model = DecisionTreeClassifier()

model.fit(X, y)

prediction = model.predict([[3.5]])

print("Prediction:", prediction)

hours = [[2], [3.5], [5]]

print("Multiple Predictions:", model.predict(hours))

print("Probability:", model.predict_proba([[3.5]]))

plot_tree(model, filled=True)
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
X = np.array([
    [22, 25],
    [25, 30],
    [28, 35],
    [35, 60],
    [40, 70],
    [45, 80],
    [50, 90]
])

y = np.array([0, 0, 0, 1, 1, 1, 1])

# X = [Age, Income]
# y = Buy?
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

new_customer = [[32, 50]]

prediction = model.predict(new_customer)

print(prediction)

prob = model.predict_proba(new_customer)

print(prob)

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))

print(confusion_matrix(y_test, y_pred))

print(model.feature_importances_)
```

### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
import numpy as np
```

```python
X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5],
    [6]
])

y = np.array([0, 0, 0, 1, 1, 1])
```

```python
model = RandomForestClassifier(n_estimators=100)

model.fit(X, y)

print(model.predict([[3.5]]))
print(model.predict_proba([[3.5]]))
```
