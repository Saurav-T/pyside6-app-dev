# Scikit Learn Tutorial

### Linear Regression

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
# X = input (features)
X = np.array([[1], [2], [3], [4], [5]])

# y = output (target)
y = np.array([2, 4, 6, 8, 10])

# Model
model = LinearRegression()

model.fit(X, y)

y_pred = model.predict([[6], [7]])

print("Predictions:", y_pred)

# Parameters learned
print("Slope (w):", model.coef_)
print("Intercept (b):", model.intercept_)
```

```python
# X = [size, rooms]
X = np.array([
    [1000, 2],
    [1500, 3],
    [2000, 3],
    [2500, 4],
    [3000, 4]
])

# y = price
y = np.array([200, 300, 400, 500, 600])

model = LinearRegression()
model.fit(X, y)

new_house = np.array([[1800, 3]])
prediction = model.predict(new_house)

print("Prediction:", prediction)
print("Coeficient:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

plt.scatter(X, y)          # real points
plt.plot(X, y_pred)        # predicted line

plt.show()
```
