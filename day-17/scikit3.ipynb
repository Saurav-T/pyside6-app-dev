### PCA

```python
import numpy as np
from sklearn.decomposition import PCA
```

```python
X = np.array([
    [170, 65],
    [172, 67],
    [175, 70],
    [180, 75],
    [185, 80]
])
# Five Samples, 2 Features
```

```python
pca = PCA(n_components=1)

X_new = pca.fit_transform(X)

print(X_new)

print(pca.explained_variance_ratio_)
print(pca.components_)
```

```python
import numpy as np
from sklearn.decomposition import PCA

# Marks of students in Maths and Physics
# Notice Students good at Math are also good at Physics.

X = np.array([
    [90, 88],
    [85, 84],
    [80, 79],
    [70, 68],
    [60, 59]
])

pca = PCA(n_components=1)

X_new = pca.fit_transform(X)

print("Reduced Data:")
print(X_new)

print("\nVariance Preserved:")
print(pca.explained_variance_ratio_)

print("\nComponents:")
print(pca.components_)
```

### Linear Regression Example

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([30, 35, 50, 60, 75])

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_)
print("Intercept:", model.intercept_)

print("Prediction for 6 years:", model.predict([[6]]))

plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.show()
```

### Logistic Regression

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

print("Class prediction:", model.predict([[3.5]]))
print("Probability:", model.predict_proba([[3.5]]))
```

### Decision Tree Example

```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier

X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

model = DecisionTreeClassifier()
model.fit(X, y)

print("Prediction:", model.predict([[3.5]]))
print("Probability:", model.predict_proba([[3.5]]))
print("Feature importance:", model.feature_importances_)
```
