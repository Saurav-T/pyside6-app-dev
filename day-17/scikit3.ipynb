# Scikit Learn Tutorial

### PCA

```python
import numpy as np
from sklearn.decomposition import PCA
```

```python
X = np.array([
    [170, 65],
    [172, 67],
    [175, 70],
    [180, 75],
    [185, 80]
])
# Five Samples, 2 Features
```

```python
pca = PCA(n_components=1)

X_new = pca.fit_transform(X)

print(X_new)

print(pca.explained_variance_ratio_)
print(pca.components_)
```
